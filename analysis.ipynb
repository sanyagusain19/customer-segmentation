```python
features=["Age","Income","total_spending","NumWebPurchases",'NumStorePurchases','NumWebVisitsMonth']
```

```python
x=df[features].copy()
```

```python
x
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x_scaled = scaler.fit_transform(x)
```

```python
x_scaled
```

```python
from sklearn.cluster import KMeans 
```

```python
wcss =[]
```

```python
for i in range(2,10):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(x_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
plt.plot(range(2,10), wcss, marker="x")
plt.title("elbow method for optimal solution")
plt.xlabel("no. of clusters ")
plt.ylabel("inertia")
```

```python
kmeans = KMeans(n_clusters = 6)
df["cluster"] = kmeans.fit_predict(x_scaled)
```

```python
df.head()
```

```python
csummary=df.groupby("cluster")[features].mean()
csummary
```

```python
df["cluster"].value_counts()
```

```python
from sklearn.decomposition import PCA
pca= PCA(n_components=2)
pca_data= pca.fit_transform(x_scaled)
df["pca1"],df["pca2"]=pca_data[:,0], pca_data[:,1]
```

```python
df["pca1"]
```

```python
sns.scatterplot(x="pca1",y="pca2",hue="cluster", data=df)
```
